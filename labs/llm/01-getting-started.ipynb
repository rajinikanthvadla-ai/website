## 2. Tiny prompt lab (no API key)

Practice structuring prompts before you wire a real model. Later labs can use Gemini / OpenAI keys you own.

```python
def build_prompt(task: str, context: str, constraints: list[str]) -> str:
    rules = "\n".join(f"- {c}" for c in constraints)
    return (
        f"Task: {task}\n\n"
        f"Context:\n{context}\n\n"
        f"Constraints:\n{rules}\n\n"
        "Return a short, direct answer."
    )

prompt = build_prompt(
    task="Explain Forward Deployed Engineering in two sentences",
    context="Audience: working software engineers evaluating a live cohort",
    constraints=[
        "No job guarantee claims",
        "Mention customer problem → production loop",
        "Keep under 60 words",
    ],
)
print(prompt)
```
